```python
df.drop_duplicates(inplace=True)
```

#Ditry values

```python
df.replace(
    ['N/A' , 'NULL' , '' ,' '] , np.nan , inplace=True
)
```

```python
df.isnull().sum()
```

#Extra spaces

```python
df = df.apply( lambda x : x.str.strip() if x.dtype == 'object' else x)
```

#Standardize text

```python
df["Country"] = df["Country"].str.title()

df["Genre"] = df["Genre"].str.title()

df["Subscription_Plan"] = df["Subscription_Plan"].str.title()

df["Account_Status"] = (df["Account_Status"].str.title())
```

```python
df['Country'].value_counts()
```

#Numeric columns

```python
numeric_col = ['Age',
               'Watch_Hours',
               'Monthly_Fee',
               'Movies_Watched',
               'Avg_Rating',
               'Monthly_Sessions',
               'Download_Count'
               ]
```

```python
for col in numeric_col:
  df[col] = pd.to_numeric(df[col] , errors='coerce')
```

```python
df.describe()
```

```python
df = df[(df['Age'] >= 18) & (df['Age']<= 100)]
```

```python
df['Age'].describe()
```

#Negative values

```python
df = df[df['Watch_Hours']>=0]
df = df[df['Monthly_Fee']>=0]
```

#Data cleaning

```python
df['Join_Date'] = pd.to_datetime(df['Join_Date'],errors='coerce')
df['Last_Watch_Date'] = pd.to_datetime(df['Last_Watch_Date'],errors='coerce')
```

```python
df[['Join_Date','Last_Watch_Date']].isnull().sum()
```

#Missing values handle

```python
df['Age'] =  df['Age'].fillna(df['Age'].mean())
df['Watch_Hours'] = df['Watch_Hours'].fillna(df['Watch_Hours'].mean() )
df['Monthly_Fee'] = df['Monthly_Fee'].fillna(df['Monthly_Fee'].mean() )
```

```python
df['Payment_Method'] = df['Payment_Method'].fillna('Unknown')
df['Device'] = df['Device'].fillna('Unknown')
df['Genre'] = df['Genre'].fillna('Unknown')
```

#Feature engineering

```python
df['Customer_Revenue'] = (df['Monthly_Fee'] * df['Monthly_Sessions'])
```

```python
df['Engagement_Score'] = (df['Watch_Hours'] + df['Monthly_Sessions'] + df['Download_Count'])
```

```python
df.head(2)
```

```python
#Watch category

bin = [0,20,50,100,np.inf]
labels = ['Low','Medium','High','Very High']

df['Watch_Category'] = pd.cut(df['Watch_Hours'],bins=bin,labels=labels)
```

```python
df.head(2)
```

```python
df['Watch_Category'] = df['Watch_Category'].cat.add_categories(['Unknown'])

df['Watch_Category'] = df['Watch_Category'].fillna('Unknown')
```

```python
df['Watch_Category'].value_counts()
```

```python
#to check missing values

df['Watch_Category'].isnull().sum()
```

```python
#customer age group

bin = [1,25,35,50,100]
label = ['18-25','26-35','36-50','51+']

df['Age_Group'] = pd.cut(df['Age'],bins=bin,labels=label)
```

```python
df.head(2)
```

```python
df['Age_Group'].value_counts()
```

```python
#Account activation flag

df['Active_Flag'] = np.where(df['Account_Status'] == 'Active',1,0)
```
